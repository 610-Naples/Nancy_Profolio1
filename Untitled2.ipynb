```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error

# 【关键】请在这里粘贴你刚才 Copy Path 得到的路径
file_path = '/cleaned_data_2018_2024.csv' # 如果在文件夹里，可能是 '/content/xxx/cleaned_data_2018_2024.csv'

df = pd.read_csv(file_path)
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

# 划分数据
split_idx = int(len(df) * 0.8)
train_data = df.iloc[:split_idx]
test_data = df.iloc[split_idx:]
```

```python
# 训练与预测
history = [x for x in train_data['Close']]
model_arima = ARIMA(history, order=(5,1,0))
model_arima_fit = model_arima.fit()
predictions_arima = model_arima_fit.forecast(steps=len(test_data))

# 计算指标
actuals = test_data['Close'].values
rmse_a = np.sqrt(mean_squared_error(actuals, predictions_arima))
mae_a = mean_absolute_error(actuals, predictions_arima)
mape_a = np.mean(np.abs((actuals - predictions_arima) / actuals)) * 100

# 画图 (Chapter 4.1 使用)
plt.figure(figsize=(12, 6))
plt.plot(test_data.index, actuals, label='Actual Price', color='blue')
plt.plot(test_data.index, predictions_arima, label='ARIMA Predicted', color='red', linestyle='--')
plt.title('ARIMA Model: AAPL Price Prediction')
plt.xlabel('Date (Year-Month)')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True)
plt.show()

print(f"ARIMA Indicator: RMSE={rmse_a:.2f}, MAE={mae_a:.2f}, MAPE={mape_a:.2f}%")
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

# 1. 归一化
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_train = scaler.fit_transform(train_data[['Close']])
scaled_test = scaler.transform(test_data[['Close']])

# 2. 滑动窗口函数 (60天)
def create_dataset(dataset, look_back=60):
    X, Y = [], []
    for i in range(len(dataset)-look_back-1):
        X.append(dataset[i:(i+look_back), 0])
        Y.append(dataset[i + look_back, 0])
    return np.array(X), np.array(Y)

X_train, y_train = create_dataset(scaled_train)
X_test, y_test = create_dataset(scaled_test)
X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], 1)
X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], 1)

# 3. 建立 LSTM 模型
model_lstm = Sequential([
    LSTM(50, return_sequences=True, input_shape=(60, 1)),
    Dropout(0.2),
    LSTM(50, return_sequences=False),
    Dropout(0.2),
    Dense(1)
])
model_lstm.compile(optimizer='adam', loss='mean_squared_error')

# 4. 训练 (Colab 下速度非常快)
model_lstm.fit(X_train, y_train, epochs=10, batch_size=32, verbose=1)

# 5. 预测与反向转换
predictions_lstm = model_lstm.predict(X_test)
predictions_lstm = scaler.inverse_transform(predictions_lstm)
actual_lstm = scaler.inverse_transform(y_test.reshape(-1, 1))

# 6. 对齐日期画图 (因为前60天没法预测，所以要对齐日期索引)
test_dates_aligned = test_data.index[61:]

plt.figure(figsize=(12, 6))
plt.plot(test_dates_aligned, actual_lstm, label='Actual Price', color='blue')
plt.plot(test_dates_aligned, predictions_lstm, label='LSTM Predicted', color='green')
plt.title('LSTM Model: AAPL Price Prediction')
plt.xlabel('Date (Year-Month)')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True)
plt.show()

# 计算 LSTM 指标
rmse_l = np.sqrt(mean_squared_error(actual_lstm, predictions_lstm))
print(f"LSTM 指标: RMSE={rmse_l:.2f}")
```

```python
# --- Model 3: Residual-based Hybrid Model (Fixed Version) ---
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt

# 1. Get ARIMA residuals from Training set
train_arima_pred = model_arima_fit.fittedvalues
train_residuals = train_data['Close'] - train_arima_pred

# 2. Prepare Residuals for LSTM (Scaling)
res_scaler = MinMaxScaler(feature_range=(0, 1))
scaled_res_train = res_scaler.fit_transform(train_residuals.values.reshape(-1, 1))

# 3. Create Sequences for Residuals (Window=60)
def create_res_seq(data, window=60):
    X, y = [], []
    for i in range(window, len(data)):
        X.append(data[i-window:i, 0])
        y.append(data[i, 0])
    return np.array(X), np.array(y)

X_res_train, y_res_train = create_res_seq(scaled_res_train)
X_res_train = X_res_train.reshape(X_res_train.shape[0], X_res_train.shape[1], 1)

# 4. Train LSTM to learn Residuals
m_hybrid = Sequential([
    LSTM(32, input_shape=(60, 1)),
    Dense(1)
])
m_hybrid.compile(optimizer='adam', loss='mse')
m_hybrid.fit(X_res_train, y_res_train, epochs=5, batch_size=16, verbose=0)

# 5. Predict Test Set Residuals
test_residuals_raw = test_data['Close'] - predictions_arima
scaled_res_test = res_scaler.transform(test_residuals_raw.values.reshape(-1, 1))

X_res_test, _ = create_res_seq(scaled_res_test) # We only need X here
X_res_test = X_res_test.reshape(X_res_test.shape[0], X_res_test.shape[1], 1)

pred_res_lstm_scaled = m_hybrid.predict(X_res_test)
pred_res_usd = res_scaler.inverse_transform(pred_res_lstm_scaled).flatten()

# 6. 【CRITICAL FIX】Aligning lengths for Final Prediction
# The LSTM prediction starts from index 60 of the test set
arima_part = predictions_arima[60:]
# Ensure they are the exact same length
min_len = min(len(arima_part), len(pred_res_usd))
final_hybrid_pred = arima_part[:min_len] + pred_res_usd[:min_len]
actual_hybrid = test_data['Close'][60:60+min_len]

# 7. Evaluate Model 3 (English Output)
rmse_h = np.sqrt(mean_squared_error(actual_hybrid, final_hybrid_pred))
mae_h = mean_absolute_error(actual_hybrid, final_hybrid_pred)
mape_h = np.mean(np.abs((actual_hybrid - final_hybrid_pred) / actual_hybrid)) * 100

print("------------------------------------------")
print("Model 3: Hybrid ARIMA-LSTM Evaluation")
print("------------------------------------------")
print(f"RMSE: {rmse_h:.4f}")
print(f"MAE : {mae_h:.4f}")
print(f"MAPE: {mape_h:.4f}%")
print("------------------------------------------")

# 8. Visualization (with Units)
plt.figure(figsize=(12, 6))
plt.plot(actual_hybrid.index, actual_hybrid, label='Actual Price', color='blue')
plt.plot(actual_hybrid.index, final_hybrid_pred, label='Hybrid Predicted', color='purple', linestyle='--')
plt.title('Model 3: Hybrid ARIMA-LSTM Prediction (AAPL)')
plt.xlabel('Date(Year-Month)')
plt.ylabel('Price (USD)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from sklearn.metrics import mean_squared_error, mean_absolute_error

print("🚀 正在运行 Model 4 滑动窗口对比实验 (Window = 20, 60, 90)...")

windows_to_test = [20, 60, 90]
results_dict = {}
predictions_dict = {}

# 创建多变量序列的函数
def create_seq(dataset, window):
    X, y = [], []
    for i in range(window, len(dataset)):
        X.append(dataset[i-window:i, :])
        y.append(dataset[i, 0])
    return np.array(X), np.array(y)

# 循环测试不同的窗口
for w in windows_to_test:
    print(f"\n--- Training with Window = {w} days ---")
    X_train_w, y_train_w = create_seq(train_scaled_m4, w)
    X_test_w, y_test_w = create_seq(test_scaled_m4, w)

    # 构建相同的 LSTM 架构进行公平对比
    m_test = Sequential([
        LSTM(64, return_sequences=True, input_shape=(X_train_w.shape[1], X_train_w.shape[2])),
        Dropout(0.2),
        LSTM(32),
        Dense(1)
    ])
    m_test.compile(optimizer='adam', loss='mse')
    m_test.fit(X_train_w, y_train_w, epochs=20, batch_size=32, verbose=0)

    # 预测与还原
    pred_res_scaled = m_test.predict(X_test_w)
    pred_res_usd = res_scaler.inverse_transform(pred_res_scaled).flatten()

    # 对齐 ARIMA 预测值和实际值
    arima_aligned = predictions_arima[w:]
    min_len = min(len(arima_aligned), len(pred_res_usd))

    final_pred = arima_aligned[:min_len] + pred_res_usd[:min_len]
    actual = test_data['Close'][w:w+min_len]

    # 记录预测结果用于画图，使用实际日期的 index
    predictions_dict[w] = pd.Series(final_pred, index=actual.index)

    # 计算指标
    rmse = np.sqrt(mean_squared_error(actual, final_pred))
    mae = mean_absolute_error(actual, final_pred)
    mape = np.mean(np.abs((actual - final_pred) / actual)) * 100

    results_dict[w] = {'RMSE': rmse, 'MAE': mae, 'MAPE': mape}
    print(f"Results for W={w} -> RMSE: {rmse:.4f}, MAE: {mae:.4f}, MAPE: {mape:.4f}%")

# ==========================================
# 画出绝美的学术对比图 (直接导出放进论文)
# ==========================================
plt.figure(figsize=(14, 7))
# 画出真实的股价走势 (作为基准)
plt.plot(test_data.index, test_data['Close'], label='Actual Price', color='black', linewidth=2)

# 定义颜色
colors = {20: 'red', 60: 'blue', 90: 'green'}

# 画出不同 Window 的预测走势
for w in windows_to_test:
    plt.plot(predictions_dict[w].index, predictions_dict[w].values,
             label=f'Predicted (Window={w})', color=colors[w], linestyle='--', alpha=0.8)

plt.title('Sensitivity Analysis: Impact of Sliding Window Size on Model 4 (AAPL)', fontsize=14)
plt.xlabel('Date (Year-Month)', fontsize=12)
plt.ylabel('Price (USD)', fontsize=12)
plt.legend(loc='best', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()

print("\n📊 终极指标对比表 (可以直接抄到论文里):")
print("Window | RMSE    | MAE     | MAPE (%)")
print("-" * 40)
for w in windows_to_test:
    res = results_dict[w]
    print(f"{w:<6} | {res['RMSE']:<7.4f} | {res['MAE']:<7.4f} | {res['MAPE']:<7.4f}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Input

# ==========================================
# 1. 加载数据
# ==========================================
df = pd.read_csv('cleaned_data_2018_2024.csv', index_col='Date', parse_dates=True)
split = int(len(df) * 0.8)
train, test = df.iloc[:split], df.iloc[split:]

# ==========================================
# 2. 归一化
# ==========================================
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df[['Close']])
train_s, test_s = scaled_data[:split], scaled_data[split:]

def create_seq(data, window=60):
    X, y = [], []
    for i in range(window, len(data)):
        X.append(data[i-window:i, 0])
        y.append(data[i, 0])
    return np.array(X), np.array(y)

X_train, y_train = create_seq(train_s)
X_test,  y_test  = create_seq(test_s)
X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], 1)
X_test  = X_test.reshape(X_test.shape[0],  X_test.shape[1],  1)

# ==========================================
# 3. 训练 Model 2: 纯 LSTM
# ==========================================
print("正在训练 Model 2: 纯 LSTM...")
m2 = Sequential([Input(shape=(60, 1)), LSTM(50), Dense(1)])
m2.compile(optimizer='adam', loss='mse')
m2.fit(X_train, y_train, epochs=10, batch_size=32, verbose=0)

# ==========================================
# 4. 反归一化 → USD 价格
# ==========================================
preds_scaled = m2.predict(X_test)
actual_usd   = scaler.inverse_transform(y_test.reshape(-1, 1)).flatten()
preds_usd    = scaler.inverse_transform(preds_scaled).flatten()

# ==========================================
# 5. 指标计算 & 打印表格 (不写进图)
# ==========================================
rmse_m2 = np.sqrt(mean_squared_error(actual_usd, preds_usd))
mae_m2  = mean_absolute_error(actual_usd, preds_usd)
mape_m2 = np.mean(np.abs((actual_usd - preds_usd) / actual_usd)) * 100

print(f"\n✅ Model 2 (LSTM) 完成!")
print("=" * 45)
print(f"{'Metric':<20} {'Value':>20}")
print("=" * 45)
print(f"{'RMSE (USD)':<20} {'$' + f'{rmse_m2:.4f}':>20}")
print(f"{'MAE  (USD)':<20} {'$' + f'{mae_m2:.4f}':>20}")
print(f"{'MAPE (%)':<20} {f'{mape_m2:.4f}%':>20}")
print("=" * 45)

# ==========================================
# 6. 可视化 (干净版，不含指标文字)
# ==========================================
def plot_lstm_predictions(test_data, actual, preds, window=60, ticker='AAPL'):
    plot_dates = test_data.index[window:]
    residuals  = actual - preds

    fig, axes = plt.subplots(2, 1, figsize=(14, 10),
                             gridspec_kw={'height_ratios': [3, 1]})

    # ── 上图：价格对比 ──────────────────────────────────────────
    ax1 = axes[0]
    ax1.plot(plot_dates, actual,
             label='Actual Close Price', color='#1f77b4',
             linewidth=1.8, zorder=3)
    ax1.plot(plot_dates, preds,
             label='LSTM Prediction', color='#d62728',
             linewidth=1.6, linestyle='--', zorder=3)
    ax1.fill_between(plot_dates, actual, preds,
                     alpha=0.12, color='#d62728', label='Prediction Error Band')

    ax1.set_title(f'Model 2 (LSTM) — Actual vs Predicted Close Price ({ticker})',
                  fontsize=14, fontweight='bold')
    ax1.set_ylabel('Price (USD)', fontsize=12)
    ax1.legend(loc='upper left', fontsize=10)
    ax1.grid(True, linestyle=':', alpha=0.6)
    ax1.tick_params(axis='x', rotation=30)

    # ── 下图：残差柱状图 ────────────────────────────────────────
    ax2 = axes[1]
    ax2.bar(plot_dates, residuals,
            color=np.where(residuals >= 0, '#2ca02c', '#d62728'),
            alpha=0.7, width=1.5)
    ax2.axhline(0, color='black', linewidth=0.9, linestyle='--')
    ax2.set_title('Prediction Residuals (Actual − Predicted)', fontsize=12)
    ax2.set_xlabel('Date (Year-Month)', fontsize=12)
    ax2.set_ylabel('Residual (USD)', fontsize=12)
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.tick_params(axis='x', rotation=30)

    plt.tight_layout()
    plt.savefig(f'model2_lstm_prediction_{ticker}.png', dpi=150, bbox_inches='tight')
    print(f"\n📁 图表已保存为 model2_lstm_prediction_{ticker}.png")
    plt.show()


plot_lstm_predictions(
    test_data = test,
    actual    = actual_usd,
    preds     = preds_usd,
    window    = 60,
    ticker    = 'AAPL'
)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input
from sklearn.metrics import mean_squared_error, mean_absolute_error
from statsmodels.tsa.arima.model import ARIMA
import shap
import os

# ==========================================
# 1. 锁定随机种子
# ==========================================
np.random.seed(42)
import tensorflow as tf
tf.random.set_seed(42)

# ==========================================
# 2. 读取数据
# ==========================================
df = pd.read_csv('cleaned_data_2018_2024.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

# ==========================================
# 3. 划分疫情数据
# ==========================================
train_df = df[:'2019-12-31']
test_df  = df['2020-01-02':'2020-06-30'].copy()

# ✅ 安全检查：确保测试集不为空
if len(test_df) == 0:
    raise ValueError("❌ test_df 为空！请检查 CSV 文件中是否包含 2020-01-02 至 2020-06-30 的数据。")

actual_prices = test_df['Close'].values
dates         = test_df.index

print(f"✅ 数据准备就绪！训练集: {len(train_df)} 天 | 疫情测试期: {len(test_df)} 天")

# ==========================================
# 统一指标打印函数 (表格风格，与前面模型统一)
# ==========================================
def print_metrics(model_name, actual, pred):
    rmse = np.sqrt(mean_squared_error(actual, pred))
    mae  = mean_absolute_error(actual, pred)
    mape = np.mean(np.abs((actual - pred) / actual)) * 100

    print(f"\n✅ {model_name} 完成!")
    print("=" * 45)
    print(f"{'Metric':<20} {'Value':>20}")
    print("=" * 45)
    print(f"{'RMSE (USD)':<20} {'$' + f'{rmse:.4f}':>20}")
    print(f"{'MAE  (USD)':<20} {'$' + f'{mae:.4f}':>20}")
    print(f"{'MAPE (%)':<20} {f'{mape:.4f}%':>20}")
    print("=" * 45)

    return rmse, mae, mape

# ==========================================
# 统一绘图函数 (风格与前面模型统一)
# ==========================================
def plot_and_save(dates, actual, pred, title, filename, ticker='AAPL'):
    residuals = actual - pred

    fig, axes = plt.subplots(2, 1, figsize=(14, 10),
                             gridspec_kw={'height_ratios': [3, 1]})

    # ── 上图：价格对比 ──────────────────────────────────────────
    ax1 = axes[0]
    ax1.plot(dates, actual,
             label='Actual Close Price', color='#1f77b4',
             linewidth=1.8, zorder=3)
    ax1.plot(dates, pred,
             label='Predicted Price', color='#d62728',
             linewidth=1.6, linestyle='--', zorder=3)
    ax1.fill_between(dates, actual, pred,
                     alpha=0.12, color='#d62728', label='Prediction Error Band')

    ax1.set_title(f'{title} ({ticker}) — COVID-19 Volatility Period',
                  fontsize=14, fontweight='bold')
    ax1.set_ylabel('Price (USD)', fontsize=12)
    ax1.legend(loc='best', fontsize=10)
    ax1.grid(True, linestyle=':', alpha=0.6)
    ax1.tick_params(axis='x', rotation=30)

    # ── 下图：残差柱状图 ────────────────────────────────────────
    ax2 = axes[1]
    ax2.bar(dates, residuals,
            color=np.where(residuals >= 0, '#2ca02c', '#d62728'),
            alpha=0.7, width=1.5)
    ax2.axhline(0, color='black', linewidth=0.9, linestyle='--')
    ax2.set_title('Prediction Residuals (Actual − Predicted)', fontsize=12)
    ax2.set_xlabel('Date (Year-Month)', fontsize=12)
    ax2.set_ylabel('Residual (USD)', fontsize=12)
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.tick_params(axis='x', rotation=30)

    plt.tight_layout()
    save_path = f"{filename}.png"
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    print(f"📁 图表已保存为 {os.path.abspath(save_path)}")
    plt.show()

# ==========================================
# Model 1: ARIMA 滚动预测
# ==========================================
print("\n🚀 正在运行 Model 1 (Pandemic): ARIMA 滚动预测...")
print(f"   预计循环 {len(test_df)} 次，请耐心等候...")

history  = list(train_df['Close'])
preds_m1 = []

for t in range(len(test_df)):
    model_arima = ARIMA(history, order=(5, 1, 0)).fit()
    preds_m1.append(model_arima.forecast()[0])
    history.append(actual_prices[t])

    # 每20步打印一次进度
    if (t + 1) % 20 == 0 or (t + 1) == len(test_df):
        print(f"   进度: {t+1}/{len(test_df)} 步完成")

preds_m1 = np.array(preds_m1)

print_metrics("Model 1 (ARIMA) — Pandemic", actual_prices, preds_m1)
plot_and_save(dates, actual_prices, preds_m1,
              "Model 1: ARIMA Prediction", "Fig_M1_Pandemic")
```

```python
print("🚀 正在运行 M2: Pure LSTM (Pandemic)...")
scaler_m2 = MinMaxScaler()
scaled_all = scaler_m2.fit_transform(df[['Close']])

# 创建 60 天窗口数据
def create_seq(data, window=60):
    X, y = [], []
    for i in range(window, len(data)):
        X.append(data[i-window:i, 0])
        y.append(data[i, 0])
    return np.array(X), np.array(y)

train_size = len(train_df)
X_train_m2, y_train_m2 = create_seq(scaled_all[:train_size])

# 训练 M2
m2 = Sequential([Input(shape=(60,1)), LSTM(50), Dense(1)])
m2.compile(optimizer='adam', loss='mse')
m2.fit(X_train_m2.reshape(-1,60,1), y_train_m2, epochs=15, verbose=0)

# 测试集准备 (需要包含前60天以预测第一天)
test_inputs = scaled_all[train_size - 60 : train_size + len(test_df)]
X_test_m2, _ = create_seq(test_inputs)

# 预测
preds_m2_scaled = m2.predict(X_test_m2.reshape(-1,60,1))
preds_m2 = scaler_m2.inverse_transform(preds_m2_scaled).flatten()

# 打印指标并画图保存
print_metrics("Model 2 (Pure LSTM)", actual_prices, preds_m2)
plot_and_save(dates, actual_prices, preds_m2, "Model 2: Pure LSTM Prediction", "Fig_M2_Pandemic")
```

```python
print("🚀 正在运行 M3: Hybrid Without Sentiment (Pandemic)...")

# 1. 计算测试集 ARIMA 的预测残差
residuals_m1 = actual_prices - preds_m1

# 2. 归一化残差
scaler_res = MinMaxScaler()
scaled_res = scaler_res.fit_transform(residuals_m1.reshape(-1, 1))

# 3. 准备 LSTM 输入 (由于测试集较短，我们简化窗口逻辑直接拟合残差趋势)
# 这里我们用前5天残差预测第6天残差，以适应短周期疫情数据
res_window = 5
X_res, y_res = [], []
for i in range(res_window, len(scaled_res)):
    X_res.append(scaled_res[i-res_window:i, 0])
    y_res.append(scaled_res[i, 0])
X_res, y_res = np.array(X_res), np.array(y_res)

m3 = Sequential([Input(shape=(res_window, 1)), LSTM(32), Dense(1)])
m3.compile(optimizer='adam', loss='mse')
m3.fit(X_res.reshape(-1, res_window, 1), y_res, epochs=20, verbose=0)

# 4. 预测残差并还原
pred_res_m3_scaled = m3.predict(X_res.reshape(-1, res_window, 1))
pred_res_m3 = scaler_res.inverse_transform(pred_res_m3_scaled).flatten()

# 5. 计算最终预测价 (ARIMA + 预测残差)
# 注意因为使用了 res_window，所以要对齐长度
final_m3 = preds_m1[res_window:] + pred_res_m3
actual_m3 = actual_prices[res_window:]
dates_m3 = dates[res_window:]

# 打印指标并画图保存
print_metrics("Model 3 (Hybrid - No Sentiment)", actual_m3, final_m3)
plot_and_save(dates_m3, actual_m3, final_m3, "Model 3: Hybrid (Price Only)", "Fig_M3_Pandemic")
```

```python
print("🚀 正在运行 M4: Sentiment Hybrid + SHAP (Pandemic)...")

# 1. 获取情感数据
sentiment_col = '1_day_sentiment_avg'
sentiment_data = test_df[sentiment_col].values

# 2. 归一化情感数据并与残差合并 (严格的双特征输入)
scaler_sent = MinMaxScaler()
scaled_sent = scaler_sent.fit_transform(sentiment_data.reshape(-1, 1))

# 特征矩阵：列1是残差，列2是情感
combined_features = np.column_stack([scaled_res, scaled_sent])

# 3. 构造双特征输入窗口
X_m4, y_m4 = [], []
for i in range(res_window, len(combined_features)):
    X_m4.append(combined_features[i-res_window:i, :])
    y_m4.append(scaled_res[i, 0]) # 目标依然是预测价格残差
X_m4, y_m4 = np.array(X_m4), np.array(y_m4)

# 4. 训练 M4
m4 = Sequential([Input(shape=(res_window, 2)), LSTM(64), Dropout(0.1), Dense(1)])
m4.compile(optimizer='adam', loss='mse')
m4.fit(X_m4, y_m4, epochs=25, verbose=0)

# 5. 预测并还原
pred_res_m4_scaled = m4.predict(X_m4)
pred_res_m4 = scaler_res.inverse_transform(pred_res_m4_scaled).flatten()

final_m4 = preds_m1[res_window:] + pred_res_m4

# 打印指标并画图保存
print_metrics("Model 4 (Sentiment Hybrid)", actual_m3, final_m4)
plot_and_save(dates_m3, actual_m3, final_m4, "Model 4: Sentiment Hybrid Prediction", "Fig_M4_Pandemic")

# ==========================================
# 6. 生成并保存 SHAP 图
# ==========================================
print("\n🔍 正在生成 SHAP 特征归因图...")
explainer = shap.GradientExplainer(m4, X_m4[:10]) # 使用部分数据作为基线
shap_values = explainer.shap_values(X_m4[10:])

shap_matrix = shap_values[0] if isinstance(shap_values, list) else shap_values
shap_2d = np.mean(shap_matrix, axis=1)
X_m4_2d = np.mean(X_m4[10:], axis=1)

plt.figure(figsize=(10, 6))
shap.summary_plot(shap_2d, X_m4_2d, feature_names=['Price Residual', 'FinBERT Sentiment'], show=False)
plt.title("SHAP Feature Importance (Pandemic Period)", fontsize=14)
plt.tight_layout()

shap_save_path = "Fig_M4_Pandemic_SHAP.png"
plt.savefig(shap_save_path, dpi=300, format='png', bbox_inches='tight')
print(f"📁 SHAP 图已保存至: {os.path.abspath(shap_save_path)}")
plt.show()
```
